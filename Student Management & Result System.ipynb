```python
# Student Information
print("--------- PERSONAL INFORMATION ---------")
name = input("Enter Name: ")
age = int(input("Enter Age: "))
city = input("Enter City: ")

# Marks
print("\n--------- MARKS INFORMATION ---------")
eng = float(input("Enter Marks of English: "))
math = float(input("Enter Marks of Mathematics: "))
phy = float(input("Enter Marks of Physics: "))
chem = float(input("Enter Marks of Chemistry: "))
pst = float(input("Enter Marks of Pakistan Studies: "))

# Total Marks
total_marks = eng + math + phy + chem + pst

# Percentage
percentage = (total_marks / 500) * 100

# Grade
if percentage >= 90:
    grade = "A+"
    status = "PASS"
elif percentage >= 80:
    grade = "A"
    status = "PASS"
elif percentage >= 70:
    grade = "B"
    status = "PASS"
elif percentage >= 60:
    grade = "C"
    status = "PASS"
elif percentage >= 50:
    grade = "D"
    status = "PASS"
elif percentage >= 40:
    grade = "E"
    status = "PASS"
else:
    grade = "Fail"
    status = "FAIL"

# Dictionary off all Subjects
all_subjects = {
    "English": eng,
    "Mathematics": math,
    "Physics": phy,
    "Chemistry": chem,
    "Pakistan Studies": pst
}

# Highest & Lowest Marks
highest_subject = max(all_subjects, key=all_subjects.get)
highest_marks = all_subjects[highest_subject]

lowest_subject = min(all_subjects, key=all_subjects.get)
lowest_marks = all_subjects[lowest_subject]

# Final Report
print("\n\n========================================")
print("        STUDENT RESULT SYSTEM")
print("========================================")

print(f"\nName     :{name}")
print(f"Age      :{age}")
print(f"City     :{city}")

print("\n----------------------------------------")
print(f"{'Subject':<21}Marks")
print("----------------------------------------")
print(f"{'English':<21}{eng}")
print(f"{'Mathematics':<21}{math}")
print(f"{'Physics':<21}{phy}")
print(f"{'Chemistry':<21}{chem}")
print(f"{'Pakistan Studies':<21}{pst}")
print("----------------------------------------")

print(f"\nTotal Marks             : {total_marks:.0f} / 500")
print(f"Percentage              : {percentage:.1f}%")
print(f"Grade                   : {grade}")
print(f"Status                  : {status}")
print()
print(f"Highest Subject         : {highest_subject}")
print(f"Highest Marks           : {highest_marks}")
print()
print(f"Lowest Subject          : {lowest_subject}")
print(f"Lowest Marks            : {lowest_marks}")

print("\n========================================")
print("             KEEP LEARNING 🚀")
print("========================================")
```
